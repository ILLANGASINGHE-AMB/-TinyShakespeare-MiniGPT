```python
import os

RAW_PATH = "../data/raw/tiny-shakespeare.txt"

PROCESSED_DIR = "../data/processed"

CLEAN_PATH = os.path.join(
    PROCESSED_DIR,
    "tiny-shakespeare_clean.txt"
)

os.makedirs(PROCESSED_DIR, exist_ok=True)

print("Raw file:", RAW_PATH)
print("Clean file:", CLEAN_PATH)
```

```python
with open(RAW_PATH, "r", encoding="utf-8") as file:
    raw_text = file.read()

print("Raw dataset loaded.")
print(f"Characters: {len(raw_text):,}")
```

```python
metadata_line = "3 KING HENRY VI"

metadata_count = raw_text.count(metadata_line)
corrupted_count = raw_text.count("sha$l")

print("Cleaning targets")
print("=" * 50)
print(f"'{metadata_line}' occurrences : {metadata_count}")
print(f"'sha$l' occurrences          : {corrupted_count}")
```

```python
raw_lines = raw_text.splitlines()

cleaned_lines = []
removed_metadata = []

for line in raw_lines:

    if line.strip() == metadata_line:
        removed_metadata.append(line)
        continue

    cleaned_lines.append(line)

print("Original lines :", len(raw_lines))
print("Cleaned lines  :", len(cleaned_lines))
print("Removed lines  :", len(removed_metadata))
```

```python
cleaned_text = "\n".join(cleaned_lines)

print("Characters after metadata removal:", len(cleaned_text))
```

```python
before_correction = cleaned_text.count("sha$l")

cleaned_text = cleaned_text.replace("sha$l", "shall")

after_correction = cleaned_text.count("sha$l")

print("Before correction:", before_correction)
print("After correction :", after_correction)
```

```python
with open(CLEAN_PATH, "w", encoding="utf-8") as file:
    file.write(cleaned_text)

print("Cleaned dataset saved successfully.")
print(CLEAN_PATH)
```

```python
cleaned_lines = cleaned_text.splitlines()

print("=" * 60)
print("DATA CLEANING REPORT")
print("=" * 60)

print(f"{'Metric':<30} {'Before':>12} {'After':>12}")
print("-" * 60)

print(
    f"{'Characters':<30}"
    f"{len(raw_text):>12,}"
    f"{len(cleaned_text):>12,}"
)

print(
    f"{'Lines':<30}"
    f"{len(raw_lines):>12,}"
    f"{len(cleaned_lines):>12,}"
)

print(
    f"{'3 KING HENRY VI':<30}"
    f"{raw_text.count(metadata_line):>12}"
    f"{cleaned_text.count(metadata_line):>12}"
)

print(
    f"{'sha$l':<30}"
    f"{raw_text.count('sha$l'):>12}"
    f"{cleaned_text.count('sha$l'):>12}"
)

print("=" * 60)
```

```python
raw_vocab = set(raw_text)
clean_vocab = set(cleaned_text)

print("Raw vocabulary size   :", len(raw_vocab))
print("Clean vocabulary size :", len(clean_vocab))

removed_characters = sorted(raw_vocab - clean_vocab)

print("\nCharacters no longer present:")
for char in removed_characters:
    print(repr(char))
```
