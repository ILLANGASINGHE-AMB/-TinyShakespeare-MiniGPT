```python
# Tiny Shakespeare — Character Tokenization

## Objective

# Convert the cleaned text corpus into numerical token IDs that can be processed by a neural network.

### Steps

# 1. Load the cleaned dataset.
# 2. Build the character vocabulary.
# 3. Create a character → integer mapping.
# 4. Create an integer → character mapping.
# 5. Implement an encoder.
# 6. Implement a decoder.
# 7. Test the encode/decode round trip.
# 8. Convert encoded data into PyTorch tensors.

# We are using character-level tokenization for this first LLM.
```

```python
import sys
import torch

print("Python:")
print(sys.executable)

print("\nPyTorch:")
print(torch.__version__)

print("\nMPS available:")
print(torch.backends.mps.is_available())
```

```python
CLEAN_PATH = "../data/processed/tiny-shakespeare_clean.txt"

with open(CLEAN_PATH, "r", encoding="utf-8") as file:
    text = file.read()

print("Clean dataset loaded.")
print(f"Characters: {len(text):,}")
```

```python
vocab = sorted(set(text))
vocab_size = len(vocab)

print("Vocabulary size:", vocab_size)
print("\nVocabulary:")
print(vocab)
```

```python
stoi = {character: index for index, character in enumerate(vocab)}

print("Character → ID")
print("=" * 40)

for character, index in stoi.items():
    print(repr(character), "→", index)
```

```python
itos = {index: character for index, character in enumerate(vocab)}

print("ID → Character")
print("=" * 40)

for index, character in itos.items():
    print(index, "→", repr(character))
```

```python
def encode(text):
    return [stoi[character] for character in text]
```

```python
sample = "Hello"

encoded = encode(sample)

print("Original:", sample)
print("Encoded :", encoded)
```

```python
def decode(tokens):
    return "".join(itos[token] for token in tokens)
```

```python
decoded = decode(encoded)

print("Encoded :", encoded)
print("Decoded :", decoded)
```

```python
sample = "To be, or not to be."

encoded = encode(sample)
decoded = decode(encoded)

print("Original:")
print(sample)

print("\nEncoded:")
print(encoded)

print("\nDecoded:")
print(decoded)

print("\nRound-trip successful:", sample == decoded)
```

```python
encoded_text = encode(text)

print("Total tokens:", len(encoded_text))
print("First 100 tokens:")
print(encoded_text[:100])
```

```python
data = torch.tensor(encoded_text, dtype=torch.long)

print("Tensor shape:", data.shape)
print("Tensor dtype:", data.dtype)
print("First 100 tokens:")
print(data[:100])
```
